# Popularity Based Recommender System

```python
ratings_with_name = ratings.merge(books, on="ISBN") #merge ratings and books
num_rating_df =ratings_with_name.groupby('Book-Title').count()['Book-Rating'].reset_index()
num_rating_df.rename(columns={'Book-Rating':'num_ratings'},inplace=True) #rename column
num_rating_df
```

```python
#checking for average rating
avg_rating_df =ratings_with_name.groupby('Book-Title').mean()['Book-Rating'].reset_index()
avg_rating_df.rename(columns={'Book-Rating':'avg_rating'},inplace=True) #rename column
avg_rating_df
```

```python
#merge num_rating_df and avg_rating df 
popular_df = num_rating_df.merge(avg_rating_df,on='Book-Title')
popular_df
```

```python
#top 50 book which is having a highest ratings
popular_df=popular_df[popular_df['num_ratings']>=250].sort_values('avg_rating',ascending=False).head(50)
popular_df
```

```python
#merge popular_df with books to get the books description
popular_df=popular_df.merge(books,on='Book-Title').drop_duplicates('Book-Title')[['Book-Title',
                                        'Book-Author','Image-URL-M','num_ratings','avg_rating']]
```

```python
popular_df
```

# Collaborative Filtering Based Recommender System

```python
#filtering on the basis of users
#filtering the dataframe : only consider the user which rate >=200 books

x=ratings_with_name.groupby('User-ID').count()['Book-Rating'] >200
users_rates=x[x].index
users_rates
```

```python
filtered_rating =ratings_with_name[ratings_with_name['User-ID'].isin(users_rates)]
```

```python
filtered_rating
```

```python
#there are only 811 users who rates more than 200 books
#474007 ratings are only done by those 811 users who rates more than 200 books
```

```python
#filtering on the basis of books
y = filtered_rating.groupby('Book-Title').count()['Book-Rating']>=50
famous_book = y[y].index
famous_book
```

```python
final_ratings = filtered_rating[filtered_rating['Book-Title'].isin(famous_book)]
final_ratings.drop_duplicates() #checking for duplicates
```

```python
#There are 58586 books which is rated by more then 50 users
```

```python
#applying pivot table
pt=final_ratings.pivot_table(index='Book-Title',columns='User-ID',values='Book-Rating')
pt.fillna(0,inplace=True) # fill 0 where NA is present
pt
```

```python
#now every book acts as a vector where we can represent it in 810 dimensions
#to find the similar vector we need euclidean distance of every vector with other vector
```

```python
#to find the euclidean distance
from sklearn.metrics.pairwise import cosine_similarity
```

```python
similarity_scores = cosine_similarity(pt)
```

```python
similarity_scores.shape
```

```python
def recommend(book_name):
    #index fetch
    index = np.where(pt.index== book_name)[0][0]
    similar_items = sorted(list(enumerate(similarity_scores[index])),key=lambda x:x[1],reverse=True)[0:10]
    
    for i in similar_items:
        print(pt.index[i[0]])
        
```

```python
def recommend(book_name):
    #index fetch
    index = np.where(pt.index== book_name)[0][0]
    similar_items = sorted(list(enumerate(similarity_scores[index])),key=lambda x:x[1],reverse=True)[0:10]
     
    data = []
    for i in similar_items:
        item = []
        temp_df = (books[books['Book-Title']==pt.index[i[0]]])
        item.extend(list(temp_df.drop_duplicates('Book-Title')['Book-Title'].values))
        item.extend(list(temp_df.drop_duplicates('Book-Title')['Book-Author'].values))
        item.extend(list(temp_df.drop_duplicates('Book-Title')['Image-URL-M'].values))
        
        data.append(item)
    return data
```
